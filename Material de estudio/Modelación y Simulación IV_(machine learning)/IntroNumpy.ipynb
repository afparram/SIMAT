# NumPy: Arreglos eficientes


Numpy (Numerical Python) es la librería fundamental de Python para realizar análisis de datos, ya que es la base sobre la cual están construidas gran parte de las otras librerías especializadas que se utilizan. Numpy provee:

- ndarray: arreglo multidimensional rápido y eficiente para operaciones numéricas. Una matriz en numpy es un arreglo de 2 dimensiones (filas y columnas). tener en cuenta que los arreglos pueden tener N-dimensiones: por ejemplo, un video puede tener coordenadas x,y de un bojeto y además una tercera dimensión del tiempo.
Un arreglo es una estructura de datos (forma de organizar los datos) con elementos homogéneos (del mismo tipo) y ordenados reconocidos por un nombre en común. 

- Permite hacer operaciones en todos los elementos sin necesidad de loops.

- Lectura y guardado de los arreglos en disco

- Sirve para álgebra lineal, generación de números aleatorios, etc.

- Permite integración con código escrito en C, C++, Fortran. Esto ha permitido muchas veces usar Python como lenguaje para migrar código de C o Fortran a una interfaz más fácil de usar como la de python.

A pesar de que muchas veces no se usará numpy directamente, conocer como funcionan los numpy arrays permitirá entender mejor qué ocurre por debajo en otras librerías (ej: pandas).

Nos enfocaremos en las siguientes funcionalidades de numpy:

- Operaciones rápidas en arreglos para limpieza, submuestreo, filtrado y transformación de datos

- Ordenamiento, indexación y operaciones entre conjuntos

- Estadística descriptiva 

- Manipulación de datos para combinar conjuntos de datos heterogéneos

- Lógica condicional usando expresiones con arreglos en vez de con loops

A pesar de que Numpy provee todas estas operaciones, es más común y sencillo usarlas con una librería como pandas, ya que provee una interfaz más amigable al usuario y fácil de explorar, además de agregar otras funcionalidades como por ejemplo el manejo de series temporales.

## El NumPy ndarray: un arreglo multidimensional

La principal característica de Numpy es su arreglo N-dimensional (ndarray), el cual es un contenedor rápido y flexible para almacenar grandes conjuntos de datos y permitir la aplicación de operaciones matemáticas en estos:

### Filtrado usando Selección Booleana

Es muy común usar filtros para seleccionar de forma condicional algunos elementos de un arreglo. Para esto se usan booleanos para realizar la selección.

Supongamos que tenemos el siguiente arreglo de nombres con duplicados:

```python
names = np.array(['Bob', 'Joe', 'Will', 'Bob', 'Will', 'Joe', 'Joe'])
data = np.random.randn(7, 4)
names
```

```python
data
```

    
Suponer que cada nombre corresponde a una fila del arreglo data y que queremos seleccionar todas las filas con el nombre 'Bob'. Usamos el operador de comparación con el arreglo:

```python
names == 'Bob'
```

Y retornamos todas las filas donde la condición sea True:

```python
data[names == 'Bob']
```

Esto se puede hacer también con negaciones:

```python
names != 'Bob' ### Donde el nombre es diferente de bob
```

```python
data[(names != 'Bob')]
```

Puedo elegir también varios nombres a incluir en mi condición:

```python
names == 'Bob'
```

```python
names == 'Will'
```

```python
### Buscar donde el nombre sea Bob o Will
mask = (names == 'Bob') | (names == 'Will') ## No se puede usar "or", sino que se debe usar |
mask
```

```python
data[mask]
```

Por ejemplo, una tarea común de limpieza de datos es cambiar todos los negativos por 0, y se usa un indice condicional sobre el arreglo (esto es muy usado para tareas de limpieza):

```python
data
```

```python
data < 0
```

```python
data[data < 0] = 0 ### Donde data sea negativo, ponerlo en 0
data
```

Nota: usar selección boolean genera copias de los arreglos, en lugar de referencias al original.

### Lógica condicional con np.where

Esto es muy común en análisis de datos. Por ejemplo, supongamos que se quieren cambiar todos los números positivos por 1 y negativos por 0:

```python
arr = np.random.randn(4, 4)
arr
```

Para lograr este objetivo, podemos usar np.where, con el arreglo de numpy que creamos, la condición que queremos (if), y que hacer si se cumple (then) y si no se cumple (else):

```python
np.where(arr > 0, 1, 0) # Las posiciones donde sea positivo, ponga 1, en las demas, ponga 0
```

Si quisiera simplemente saber en que posiciones se cumple y en cuales no la condición, podría hacer simplemente esto:

```python
(arr > 0)
```

### Ordenamiento

Al igual que las listas, podemos usar sort para ordenar un arreglo. El sort se hace inplace:

```python
arr = np.random.randn(6)
arr
```

```python
arr.sort()
arr
```

```python
arr[::-1] # este es un truco que se hace para ponerlo en el orden contrario
```

Para arreglos multidimensionales, puedo especificar sobre que dimensión se ordenará:

```python
arr = np.random.randn(5, 3)
arr
```

```python
arr.sort(1) ## sobre las filas es 1, sobre las columnas es 0
arr
```

## Álgebra lineal



Numpy permite hacer varias operaciones de álgebra lineal: multiplicación matricial, descomposiciones, determinantes, etc. Multiplicar en Numpy con * es elemento a elemento; si se quiere multiplicación matricial, usar np.dot:

```python
x = np.array([[1., 2., 3.], [4., 5., 6.]])
y = np.array([[6., 23.], [-1, 7], [8, 9]])
x
```

```python
y
```

```python
x.dot(y)
```

```python
np.dot(x, y)
```

```python
### Se puede usar también @
x@y
```

```python
np.dot(x, np.ones(3))
```

Numpy.linalg tiene varias operaciones también muy comunes en álgebra lineal (como por ejemplo la inversa y la traspuesta). En el siguiente ejercicio, calculamos $(X^TX)^{-1}$. Esta operación es muy común en álgebra matricial y se usa para la solución de la regresión lineal múltiple por mínimos cuadrados ordinarios. (Se calcula la transpuesta de la matriz X, se multiplica por X y se invierte la matriz resultado).

```python
from numpy.linalg import inv
X = np.random.randn(5, 5)
Xt = X.T
resultado_mult = Xt.dot(X)
resultado_final = inv(resultado_mult)
resultado_final

```

```python
# en una linea
inv(X.T.dot(X))
```
