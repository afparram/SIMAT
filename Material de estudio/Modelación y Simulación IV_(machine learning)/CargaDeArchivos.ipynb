El formato json es mucho más libre que el csv.


Hay muchas librerías para trabajar con json. Una es la librería "json": con json.loads podemos convertir un JSON string a un diccionario. Para la operación inversa, se usa json.dumps.

Si le pasamos una lista de json a un objeto Dataframe, se construirá una dataframe usando todos esos json, siempre y cuando todos los objetos tengan la misma longitud.


```python
obj = """
{"name": "Wes",
 "places_lived": ["United States", "Spain", "Germany"],
 "pet": null,
 "siblings": [{"name": "Scott", "age": 30, "pets": ["Zeus", "Zuko"]},
              {"name": "Katie", "age": 38,
               "pets": ["Sixes", "Stache", "Cisco"]}]
}
"""
```

```python
obj
```

```python
import json

result = json.loads(obj)
result
```

```python
type(result)
```

```python
asjson = json.dumps(result)
asjson
```

```python
type(asjson)
```

```python
### Si elijo el campo siblings del json, me permite convertirlo a dataframe
siblings = pd.DataFrame(result['siblings'])
siblings
```

## Formatos binarios

La forma más fácil de almacenar objetos de python de forma binaria es mediante "serialización", usando la librería pickle. La forma binaria es muy "cruda" y pura para el computador, por lo que será mas eficiente (menos espacio). Pero para nosotros no sera entendible (intentar abrir un pickle con notepad.).

Podemos guardar un objeto de pandas como un pickle (binario) usando to_pickle, y leerlo usando read_pickle. Tener en cuenta que, según la versión de pickle que se use, un objeto podría luego poner problemas para cargarse a futuro, si la versión de pickle cambia.

```python
frame = pd.read_csv('examples/ex1.csv')
frame
```

```python
frame.to_pickle('examples/frame_pickle')
```

```python
pd.read_pickle('examples/frame_pickle')
```

```python
datafp = pd.read_pickle('examples/frame_pickle')
datafp
```

```python
from sklearn.linear_model import LinearRegression



```

```python
reg_lineal = LinearRegression()
```

```python
X= datafp[['a','b','c']]
X
```

```python
y = datafp['d']
y
```

```python
reg_lineal.fit(X,y)
```

```python
reg_lineal.predict(X)
```

```python
# save the model to disk
import pickle
filename = 'examples/finalized_model.sav'
pickle.dump(reg_lineal, open(filename, 'wb'))
```

```python
# load the model from disk
loaded_model = pickle.load(open(filename, 'rb'))
```

```python
loaded_model.predict(X)
```

### Bases de datos

Las bases de datos relacionales basadas en SQL son muy comunes para almacenar datos (SQL Server, MySQL, etc).

Cargar datos desde una base de datos SQL a un dataframe es sencillo, pandas nos permite hacerlo sin mayor problema.

Vamos a crear una base de datos SQLite usando el driver sqlite3 de Python.

```python
import sqlite3

import time
nombre_tabla_unico = "test" + str(int(time.time())  )

# Notar que uso una f-string para reemplazar el valor de la variable en la string
query = f"""
CREATE TABLE {nombre_tabla_unico}
(a VARCHAR(20), b VARCHAR(20),
c REAL, d INTEGER);
"""
### declaro Dos variables varchar, una real (float), y una entera)

## Y establezco la conexiona a la base de datos mydata para crear la tabla test1

con = sqlite3.connect('examples/mydata.sqlite')
con.execute(query)
con.commit()
```

```python
### Agrego 4 filas a la tabla con un query
data = [('Atlanta', 'Georgia', 1.25, 6),
        ('Tallahassee', 'Florida', 2.6, 3),
        ('Albany', 'New York', 3, 2),
        ('Sacramento', 'California', 1.7, 5)]

stmt = f"INSERT INTO {nombre_tabla_unico} VALUES(?,?,?,?)"

con.executemany(stmt, data)
```

```python
# commit para confirmar los querys y cambios realizados
con.commit()
```


La mayoria de los drivers retornan una lista de tuplas cuando uso un Select.

Podemos usar estas listas de tuplas para el Dataframe.

```python
### Traer todas las filas
cursor = con.execute('select * from ' + nombre_tabla_unico)
rows = cursor.fetchall()
rows
```

```python
pd.DataFrame(rows, columns=[x[0] for x in cursor.description])
```


Para ahorrarnos gran parte de estas lineas de codigo (la conexion, los commit, etc), podemos usar SQLAlchemy, la cual concilia varias diferencias entre bases de datos SQL. Con esta librería, podemos luego usar pandas.read_sql para leer directamente la query que queramos y generar la tabla.


```python
import sqlalchemy as sqla

db = sqla.create_engine('sqlite:///examples/mydata.sqlite')
tabla=pd.read_sql('select * from ' + nombre_tabla_unico, db)
```

```python
tabla
```

Por ejemplo, el siguiente ejemplo muestra como podemos usar un query de SQL para traer el estado cuya capital es Tallahassee.


Note las 3 partes principales del query de SQL: el "Select" (que sirve para ver que columnas queremos traer), el "FROM" (que referencia la tabla de la cual queremos extraer los datos), y el WHERE (que indica cuáles con las condiciones que queremos que cumplan los datos que traeremos).


```python
# Aqui construyo el query de SQL que quiero ejecutar, usando el nombre de la tabla que había creado (el
# cual esta guardado en la variable "nombre_tabla_unico")
query_ejemplo = "SELECT b FROM " + nombre_tabla_unico + " WHERE a='Tallahassee';"
print(query_ejemplo)
```

```python
# Ejecutar el query con  la conexion a la base de datos y mostrar el resultado
tabla=pd.read_sql(query_ejemplo, db)
tabla
```
